```python
import pandas as pd
import numpy as np
import joblib
import json
from sklearn.preprocessing import LabelEncoder
```

```python
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, accuracy_score
```

```python
dataset = pd.read_csv("/content/drive/MyDrive/headache1.csv")
dataset.head()
```

```python
X = dataset.drop('Headache_Type', axis=1)
y = dataset['Headache_Type']

X_encoded = pd.get_dummies(X, drop_first=False)
X_encoded = X_encoded.astype(int)

le = LabelEncoder()

y_encoded = le.fit_transform(y)

target_mapping = dict(zip(le.classes_, le.transform(le.classes_)))

display(X_encoded.head())
display(y_encoded)
print(target_mapping)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X_encoded, y_encoded, test_size=0.2, random_state=42, stratify=y_encoded
)

rf_model = RandomForestClassifier(n_estimators=1000, random_state=42)

rf_model.fit(X_train, y_train)

y_pred = rf_model.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)
print("Overall Model Accuracy:", accuracy)

print("Detailed Classification Report:")
print(classification_report(y_test, y_pred, target_names=le.classes_))
```

```python
rf_base = RandomForestClassifier(random_state=42)

param_grid = {
    'n_estimators': [50, 100, 200, 300],
    'max_depth': [None, 5, 10, 15, 20],
    'max_features': ['sqrt', 'log2', None]
}

grid_search = GridSearchCV(
    estimator=rf_base,
    param_grid=param_grid,
    cv=5,
    n_jobs=-1,
    verbose=2
)

print("Dont interrupt cell runtime...")
grid_search.fit(X_train, y_train)

#print("Accuracy score(mean test score) of All Combinations:")
#means = grid_search.cv_results_['mean_test_score']
#params = grid_search.cv_results_['params']

#for mean_score, parameters in zip(means, params):
    #print("Accuracy:",mean_score, "Parameters:", parameters)

print("Display Best set of parameter combination")
print("Best Parameter:",grid_search.best_params_)
print("Best Cross-Validation Accuracy:",grid_search.best_score_)

best_rf_model = grid_search.best_estimator_
tuned_acc = best_rf_model.score(X_test, y_test)

print("Final Tuned Test Accuracy:",tuned_acc)
```
